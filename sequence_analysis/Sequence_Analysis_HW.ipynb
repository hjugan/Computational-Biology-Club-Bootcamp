Use:

\[
GC\% = \frac{G + C}{\text{sequence length}} \times 100
\]

Try to calculate it yourself using Python rather than a pre-built GC function.


```python
sequence = record.seq
counter = 0
print(type(sequence))
print(type(sequence[i]))
for i in range(92):
  if sequence[i] == 'G':
    counter += 1
  elif sequence[i] == 'C':
    counter += 1

print(counter)

percent_GC = counter / len(sequence) * 100
print(f"{percent_GC}%")

# Count G bases


# Count C bases


# Calculate GC percentage


# Print the result
```
