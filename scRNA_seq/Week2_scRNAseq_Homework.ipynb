## 0F — Optional CS Challenge: Linked Lists

A Python `list` is **not** a linked list.

A linked list stores values inside connected **nodes**:

```text
Node → Node → Node → None
```

Each node stores:
1. a value
2. a reference to the next node

This is a useful CS concept, but it is **not needed for scRNA-seq homework**.


````python
class Node:
    def __init__(self, value):
        self.value = value
        self.next = None

first = Node("Gene A")
second = Node("Gene B")
third = Node("Gene C")

first.next = second
second.next = third

current = first

while current is not None:
    print(current.value)
    current = current.next
````

## 0G — Optional CS Challenge: Recursion

A **recursive function calls itself**.

It must have a **base case** so that it eventually stops.

Example:


````python
def countdown(n):
    if n == 0:      # base case
        print("Done!")
        return

    print(n)
    countdown(n - 1)

countdown(5)
````

### Optional recursion challenge
Complete the recursive factorial function.

Remember:

```text
4! = 4 × 3 × 2 × 1
```


````python
def factorial(n):
    if n == 0:
        return 1
    
    return n * factorial(n-1)

print(factorial(4))   # should print 24
````

# Problem 2 — Quality Control

Cells can differ in:
- total RNA counts
- number of detected genes
- mitochondrial RNA percentage

These are common QC measurements.

### Helpful reading
- [Scanpy QC metrics documentation](https://scanpy.readthedocs.io/en/stable/generated/scanpy.pp.calculate_qc_metrics.html)
- [Scanpy preprocessing tutorial](https://scanpy.readthedocs.io/en/stable/tutorials/basics/clustering.html)


````python
# Total counts per cell
total_counts = counts.sum(axis=1)

# Number of genes detected (>0) per cell
n_genes = (counts > 0).sum(axis=1)

# Mitochondrial genes in our toy dataset start with "MT-"
mt_genes = [g for g in counts.columns if g.startswith("MT-")]

mt_counts = counts[mt_genes].sum(axis=1)
pct_mt = mt_counts / total_counts * 100

qc = pd.DataFrame({
    "total_counts": total_counts,
    "n_genes": n_genes,
    "pct_mito": pct_mt
})

qc.head()
````

````python
# Plot total counts
plt.figure(figsize=(7,4))
plt.hist(qc["total_counts"], bins=15)
plt.xlabel("Total counts per cell")
plt.ylabel("Number of cells")
plt.title("Total Counts per Cell")
plt.show()

# Plot mitochondrial percentage
plt.figure(figsize=(7,4))
plt.hist(qc["pct_mito"], bins=15)
plt.xlabel("Mitochondrial RNA (%)")
plt.ylabel("Number of cells")
plt.title("Mitochondrial Percentage")
plt.show()
````

### Q2 — Interpret QC

1. Which cells have unusually low total counts?  **ANswer:** Cells 5 and 27
2. Which cell has unusually high mitochondrial percentage? **Answer:** Cell 44
3. Why might these cells be considered lower quality? Having a low number of RNA molecules indicates that the cell is broken, or perhaps there is no cell at all, and just ambient RNA. If the cell broke, it is possible that all the RNA leaked out.  
4. Why can a cell with unusually high counts sometimes be suspicious? It can indicate that there is a very low amount of RNA without mitochondria. This can indicate that the cell is broken because the RNA in the cytoplasm could have leaked out, leaving only the mitochondrial RNA behind.


````python
print(f"mean of total counts: {np.mean(total_counts)}\n std of total_counts: {np.std(total_counts)}")

tc_outlier = []
for cell_id, value in qc["total_counts"].items():
    if value < 12:
        tc_outlier.append(cell_id) # assign the ID to the list
        
print(f"Cells w/ total count < 12 (3 STD below mean): {tc_outlier}")

print(f"mean of pct_mnito: {np.mean(pct_mt)}\n std of pct_mito: {np.std(pct_mt)}")


pmito_high = []
for cell_id, value in qc["pct_mito"].items():
    if value > 39:
        pmito_high.append(cell_id)

print(f"Cells w/ percent mitochondria > 39% (3 STD above mean): {pmito_high}")     



"""# Helpful: sort cells by QC metrics
print("Lowest total counts:")
print(qc.sort_values("total_counts").head())

print("\nHighest mitochondrial percentage:")
print(qc.sort_values("pct_mito", ascending=False).head())
"""
````

# Problem 4 — Normalize Gene Expression

Cells can have different sequencing depths.

We will normalize each cell to the same total count and then apply a log transform.

### Helpful reading
- [Scanpy normalization documentation](https://scanpy.readthedocs.io/en/stable/generated/scanpy.pp.normalize_total.html)
- [NumPy log1p](https://numpy.org/doc/stable/reference/generated/numpy.log1p.html)


````python
# Normalize each cell to a total of 10,000 counts
normalized = filtered_counts.div(filtered_counts.sum(axis=1), axis=0) * 10000

# Log transform
log_norm = np.log1p(normalized) #log is base e!!

print(log_norm.head())
````

### Q4 Short Answer

1. Why do we normalize cells before comparing them? Normalizing the cells before comparing them transforms the data into a proportionality problem. The question we are asking now is "what proportion of the cell's RNA belongs to this gene?" This can make it easier for evaluating and classifying certain properties of the data set
2. What problem would arise if one cell had 5× more sequencing reads than another? If the cells were not normalized, then the data would be disproportionate and difficult to analyze. A cell with 5x more sequencing reads would have way more apparent genes, however this would just be an artifact of the additional measurements. It could lead to a missclassification of the cell. This problem is omitted when we normalize the data, because it gives us the percentage of present genes, which essentially treats every cell as if it has been sequenced the same amount of times.

# Problem 5 — PCA

Each cell is described by many gene-expression values.

PCA compresses those measurements into a smaller number of major patterns.

### Helpful reading
- [scikit-learn PCA documentation](https://scikit-learn.org/stable/modules/generated/sklearn.decomposition.PCA.html)
- [PCA explained in the scikit-learn user guide](https://scikit-learn.org/stable/modules/decomposition.html#pca)


````python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
pcs = pca.fit_transform(log_norm)

pca_df = pd.DataFrame(
    pcs,
    index=log_norm.index,
    columns=["PC1", "PC2"]
)

print("Variance explained:", pca.explained_variance_ratio_)
pca_df.head()
````

````python
plt.figure(figsize=(7,5))
plt.scatter(pca_df["PC1"], pca_df["PC2"])
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA of Cells")
plt.show()
````

````python
print(pca_df.sort_values("PC1").head(10))
print(pca_df.sort_values("PC2").head(10))
print(pca_df.sort_values("PC1", ascending=False).head(10))
print(pca_df.sort_values("PC2", ascending=False).head(10))
print(pca_df.shape)
````

````python
diffs = []
n = len(pca_df)

for i in range(n):
    for j in range(i + 1, n): # j > i, so each pair is counted once
        dx = pca_df["PC1"].iloc[i] - pca_df["PC1"].iloc[j]
        dy = pca_df["PC2"].iloc[i] - pca_df["PC2"].iloc[j]
        dist = np.sqrt(dx**2 + dy**2) # magnitude of the distance b/w the points in PCA space
        diffs.append((pca_df.index[i], pca_df.index[j], dist))

diffs_df = pd.DataFrame(diffs, columns=["cell_a", "cell_b", "distance"])
print(diffs_df.sort_values("distance").head(10))
````

````python
plt.figure(figsize=(7,5))
plt.scatter(umap_df["UMAP1"], umap_df["UMAP2"])
plt.xlabel("UMAP1")
plt.ylabel("UMAP2")
plt.title("UMAP of Cells")
plt.show()
````

# Problem 7 — Clustering

We will use **K-means** to group cells based on normalized expression.

You do **not** need to understand the full K-means algorithm yet.

For this toy dataset, use **3 clusters**.


````python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42, n_init=20)
clusters = kmeans.fit_predict(log_norm)

umap_df["cluster"] = clusters.astype(str)
pca_df["cluster"] = clusters.astype(str)

print(umap_df["cluster"].value_counts())
````

````python
plt.figure(figsize=(7,5))

for cluster in sorted(umap_df["cluster"].unique()):
    subset = umap_df[umap_df["cluster"] == cluster]
    plt.scatter(
        subset["UMAP1"],
        subset["UMAP2"],
        label=f"Cluster {cluster}"
    )

plt.xlabel("UMAP1")
plt.ylabel("UMAP2")
plt.title("UMAP Colored by Cluster")
plt.legend()
plt.show()
````

# Problem 8 — Marker Genes

Now we want to determine what each cluster might represent biologically.

For each cluster, calculate the **average expression of each gene**.


````python
cluster_labels = pd.Series(clusters, index=log_norm.index, name="cluster")

cluster_means = log_norm.groupby(cluster_labels).mean()

cluster_means
````

````python
# Print the top 4 genes in each cluster
for cluster in cluster_means.index:
    top_genes = cluster_means.loc[cluster].sort_values(ascending=False).head(4)

    print(f"\nCluster {cluster}")
    print(top_genes)
````
